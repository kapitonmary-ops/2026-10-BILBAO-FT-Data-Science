## Ejercicio 12
¿Existe alguna relación lineal entre las variables numéricas del dataset? Intenta pintar este razonamiento mediante un gráfico

```python
beer_num = beer.select_dtypes(include='number')

correlacion = beer_num.corr()
display(correlacion)
#select_dtypes(include='number') — оставляет числовые столбцы.
#.corr() — рассчитывает коэффициенты корреляции Пирсона между ними.
# display() — показывает полученную таблицу.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(correlacion, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlación entre variables numéricas')
plt.show()
'''La matriz de correlación permite identificar las relaciones lineales entre las variables numéricas. 
Los coeficientes cercanos a 1 o −1 indican relaciones lineales fuertes, 
mientras que los cercanos a 0 indican relaciones débiles.'''

# annot=True — показывает числовые значения корреляций.
# cmap='coolwarm' — использует цветовую шкалу для положительных и отрицательных связей.
# vmin=-1, vmax=1 — фиксирует границы шкалы корреляции.
```
