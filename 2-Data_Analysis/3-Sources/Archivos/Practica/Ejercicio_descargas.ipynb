- Otros

```python
import os
import shutil
```

```python
# Definimos la ruta de la carpeta
ruta = os.path.expanduser("~/Desktop/Bridge(copy)")
```

```python
# Obtenemos la lista de archivos de la carpeta
archivos = os.listdir(ruta)

# Mostramos la lista
archivos
```

```python
# Recorremos la lista de archivos
for archivo in archivos:

    # Obtenemos la extensión del archivo
    extension = os.path.splitext(archivo)[1]

    # Mostramos el archivo y su extensión
    print(archivo, extension)
    
    
```

```python
# Definimos las categorías de archivos
categorias = {
    'Documentos': ['.doc', '.docx', '.txt', '.pdf', '.xls', '.ppt', '.xlsx', '.pptx'],
    'Imagenes': ['.jpg', '.jpeg', '.png', '.svg', '.gif'],
    'Software': ['.exe', '.py','.ipynb'],
    'Otros': []
}
```

```python
# Recorremos las categorías y sus extensiones
for categoria, extensiones in categorias.items():

    # Recorremos los archivos de la carpeta
    for archivo in os.listdir(ruta):

        # Recorremos las extensiones de cada categoría
        for extension in extensiones:

            # Comprobamos si el archivo tiene esta extensión
            if archivo.endswith(extension):

                # Mostramos la categoría y el archivo
                print(categoria, archivo)
```

```python
#Recorremos las categorías
for categoria in categorias:

    # Creamos la ruta de la carpeta de la categoría
    ruta_categoria = os.path.join(ruta, categoria)

    # Comprobamos si la carpeta existe
    if not os.path.exists(ruta_categoria):
        
        # Creamos la carpeta
        os.makedirs(ruta_categoria)
```

```python
# Definimos la ruta del archivo
archivo = os.path.join(ruta, 'Chuleta_pandas.pdf') #  формирует полный путь к файлу.

# Definimos la ruta de la carpeta de destino
destino = os.path.join(ruta, 'Documentos')# формирует путь к папке назначения.

# Movemos el archivo
shutil.move(archivo, destino)# перемещает файл.
```

```python
# Comprobamos el contenido de la carpeta Documentos
os.listdir(os.path.join(ruta, 'Documentos'))
```

```python
# Recorremos los archivos de la carpeta
for archivo in os.listdir(ruta):

    # Obtenemos la extensión del archivo
    extension = os.path.splitext(archivo)[1]

    # Comprobamos si el archivo es un directorio
    if os.path.isdir(os.path.join(ruta, archivo)):
        continue

    # Buscamos la categoría correspondiente
    for categoria, extensiones in categorias.items():

        # Comprobamos si la extensión pertenece a la categoría
        if extension in extensiones:

            # Definimos la ruta de origen
            origen = os.path.join(ruta, archivo)

            # Definimos la ruta de destino
            destino = os.path.join(ruta, categoria)

            # Movemos el archivo
            shutil.move(origen, destino)

            # Terminamos la búsqueda de categoría
            break

    # Si el archivo no pertenece a ninguna categoría
    else:
        destino = os.path.join(ruta, 'Otros')

        # Movemos el archivo a Otros
        shutil.move(os.path.join(ruta, archivo), destino)
```

```python
# Comprobamos el contenido de la carpeta
os.listdir(ruta)
```

```python
# Comprobamos los archivos de cada categoría
for categoria in categorias:
    print(categoria, os.listdir(os.path.join(ruta, categoria)))
```

```python
# Definimos una función para crear las carpetas
def crear_carpetas():

    # Recorremos las categorías
    for categoria in categorias:

        # Definimos la ruta de la carpeta
        ruta_categoria = os.path.join(ruta, categoria)

        # Comprobamos si la carpeta existe
        if not os.path.exists(ruta_categoria):

            # Creamos la carpeta
            os.mkdir(ruta_categoria)
```

```python
# Definimos una función para organizar los archivos
def organizar_archivos():

    # Recorremos los archivos de la carpeta
    for archivo in os.listdir(ruta):

        # Comprobamos si el elemento es una carpeta
        if os.path.isdir(os.path.join(ruta, archivo)):
            continue

        # Obtenemos la extensión del archivo
        extension = os.path.splitext(archivo)[1]

        # Buscamos la categoría correspondiente
        for categoria, extensiones in categorias.items():

            # Comprobamos si la extensión pertenece a la categoría
            if extension in extensiones:

                # Definimos la ruta de origen
                origen = os.path.join(ruta, archivo)

                # Definimos la ruta de destino
                destino = os.path.join(ruta, categoria)

                # Movemos el archivo
                shutil.move(origen, destino)

                # Terminamos la búsqueda
                break

        # Si el archivo no pertenece a ninguna categoría
        else:

            # Definimos la ruta de destino
            destino = os.path.join(ruta, 'Otros')

            # Movemos el archivo
            shutil.move(os.path.join(ruta, archivo), destino)
```

```python
# Definimos la función principal
def main():

    # Creamos las carpetas
    crear_carpetas()

    # Organizamos los archivos
    organizar_archivos()
```

```python
# Ejecutamos el programa principal
main()
```

```python
# Comprobamos el resultado
for categoria in categorias:
    print(categoria, os.listdir(os.path.join(ruta, categoria)))
```
