```python
import pandas as pd
import matplotlib.pyplot as plt

levels = ["ED0-2", "ED3_4", "ED5-8"]

level_names = {
    "ED0-2": "Bajo (ED0-2)",
    "ED3_4": "Medio (ED3-4)",
    "ED5-8": "Alto (ED5-8)"
}

colors = {
    "FBO": "#ff7f00",
    "NBO_NAT": "#00a878"
}

years = trend_eu.index.astype(int)
```

```python
# Gráfico 1: Evolución de la tasa de empleo

fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey=True)

for ax, lvl in zip(axes, levels):

    fbo = trend_eu[(lvl, "FBO")]
    nat = trend_eu[(lvl, "NBO_NAT")]

    ax.plot(
        years, fbo,
        marker="o",
        color=colors["FBO"],
        linewidth=2.5,
        label="Migrantes (FBO)"
    )

    ax.plot(
        years, nat,
        marker="o",
        color=colors["NBO_NAT"],
        linewidth=2.5,
        label="No migrantes (NBO_NAT)"
    )

    ax.fill_between(
        years, fbo, nat,
        color="#b0b0b0",
        alpha=0.2
    )

    ax.set_title(level_names[lvl])
    ax.set_xticks(years)
    ax.grid(alpha=0.3)

axes[0].set_ylabel("Tasa de empleo (%)")
axes[0].legend(loc="best")

fig.suptitle(
    "Evolución de la tasa de empleo en la UE27",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```
