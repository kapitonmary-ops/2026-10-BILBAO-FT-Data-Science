## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
#Создаем функцию

def ahorcado ():
    word = "Літера" 
    errors = 0  #указываем кол-во ошибок 0, чтобы считать и записывать кол-во ошибок
    print ("Let`s begin!")
    
#создаем пустое слово с пробелами, что видит игрок
    user_word = []
    for letter in range (len(word)): #считает кол-во букв в загаданном слове
        user_word.append("_") #выводит вместо букв символ что указан в скобках
    print (" ".join (user_word))

# Повторяем ходы, пока ошибок меньше 6
    while errors < 6:
# Проверяем букву и открываем совпадения
        user_letter = input("Tell me a letter").lower() #чтоб регистр букв не учитывался
        for i, letter1 in enumerate(word): #проверяем каждую букву в слове  
            if user_letter == letter1.lower():
                user_word[i] = letter1 # ставим букву на место пробела, что соотв загаданному слову

# Чтоб выйти из игры, когда захочешь
        else: 
                user_letter == input ("Enter 1 to stop a game")
                print ("Exit")
            
                break
            
# Если буквы нет, увеличиваем ошибки
        if user_letter not in word: 
            errors +=1
# Показываем текущее слово
        print (" ".join(user_word))
        if user_word == list(word):   
            print ("You WIN!")

            break
        
# Считаем оставшиеся попытки   
        attempts_left= 6 - errors
        if attempts_left == 0:
            print ("Game over!") 
        else: 
            print ("Attempts left ", attempts_left )
    
ahorcado ()
````
