## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
operación = input (
    "Elige operación que qieres realizar : +, /, *, //, - , **, historial, salir")

while operación != "salir":
    numero1 = int(input("Indica numero1"))
    numero2 = int(input("Indica numero2"))
    if operación == "+":
        resultado_sum = (numero1 + numero2)
        print (resultado_sum)
    elif operación == "/":
        if numero2 != 0:
            resultado_div = (numero1 / numero2)
            print (resultado_div)
        else: 
            print ("No se puede dividir por cero")
    elif operación == "*":
        resultado_mult = (numero1 * numero2)
        print (resultado_mult)
    elif operación == "//":
        resultado_rest = (numero1 // numero2)
    elif operación == "**":
        resultado_m = (numero1 ** numero2)
        print (resultado_m)
    elif operación == "-":
        resultado_r = (numero1 - numero2)
        print (resultado_r)
    
    


```
