## Ejercicio 4
Realiza un conversor de grados Celsius a Fahrenheit. El conversor tiene que recibir el input del usuario y devolver la conversión.

La conversión de grados sigue la siguiente fórmula: F = C × (9 / 5) + 32 

```python
celsius = float(input("Introduce la temperatura en grados Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(fahrenheit)
```

```python
"""
Conversor de grados Celsius a Fahrenheit.
El usuario introduce la temperatura en grados Celsius y el programa calcula su equivalente en grados Fahrenheit.
"""
# Pedimos al usuario que introduzca la temperatura en grados Celsius
celsius = float(input("Introduce la temperatura en grados Celsius: "))

# Aplicamos la fórmula para convertir a grados Fahrenheit
fahrenheit = (celsius * 9/5) + 32

# Mostramos el resultado
print(fahrenheit)
```
