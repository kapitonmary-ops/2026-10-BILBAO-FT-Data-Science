## Ejercicio 1
Dada la siguiente lista:
> ```ejer_1 = [1,2,3,4,5]```

Inviértela para que quede de la siguiente manera
> ```ejer_1 = [5,4,3,2,1]```

````python
ejer_1 = [1,2,3,4,5]
print (ejer_1.reverse())
print (ejer_1)
````

Otra forma

````python
ejer_1 = [1,2,3,4,5]
ejer_2 = list(reversed(ejer_1)) 
print (ejer_2)
````

Otra forma

````python
ejer_1 = [1,2,3,4,5]
ejer_2 = []
for numero in reversed (ejer_1):
    ejer_2.append(numero)
print(ejer_2)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
ejer_5_copy = ejer_5.copy()
for number in ejer_5_copy:
    if number ==3:
        ejer_5.remove(number)
print (ejer_5)
  

       
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]
my_dict1 = {}
for clave, valor in (ejer_11):
    my_dict1 [clave] = valor 

print (my_dict1)
````
